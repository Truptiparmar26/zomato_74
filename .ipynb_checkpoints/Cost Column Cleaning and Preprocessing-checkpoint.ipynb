```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('zomato.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
cost_col = 'approx_cost(for two people)'
```

```python
df[cost_col].head(10)
```

```python
df[cost_col].dtype
```

```python
df[cost_col].isnull().sum()
```

```python
df[cost_col].unique()[:20]
```

```python
df[cost_col] = df[cost_col].astype('string')
```

```python
df[cost_col] = df[cost_col].str.replace(',', '', regex=False)
```

```python
df[cost_col] = df[cost_col].str.replace(r'[^0-9.]', '', regex=True)
```

```python
df[cost_col] = pd.to_numeric(df[cost_col], errors='coerce')
```

```python
df[cost_col].dtype
```

```python
df[cost_col] = df[cost_col].fillna(df[cost_col].median())
```

```python
print("Missing Values:", df[cost_col].isnull().sum())

print("Data Type:", df[cost_col].dtype)

print("Minimum Cost:", df[cost_col].min())

print("Maximum Cost:", df[cost_col].max())
```

```python
df[[cost_col]].head(10)
```

```python
df.to_csv('zomato_cost_cleaned.csv', index=False)
```
